```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOpenAI(model="gpt-4o")

# 프롬프트 템플릿 생성
prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 머신러닝 전문가입니다. 한국어로 답변해주세요."),
    ("human", "{question}")
])

# 체인 구성 (LCEL)
chain = prompt | llm

# 실행
response = chain.invoke({"question": "딥러닝과 머신러닝의 차이는?"})
print(response.content)
```

# batch 처리

```python
llm = ChatOpenAI(model="gpt-4o")

# 여러 질문을 한 번에 처리
questions = [
    "파이썬이란?",
    "자바스크립트란?",
    "머신러닝이란?"
]

# batch 메서드 사용
responses = llm.batch(questions)
```

```python
# batch 처리 결과 출력
for i, response in enumerate(responses):
    print(f"\n질문 {i+1}: {questions[i]}")
    print(f"답변: {response.content}\n")
```

# Async vs Sync

```python
llm = ChatOpenAI(
    model="gpt-4o", 
    api_key=api_key
)
messages = [
    ("system", "당신은 서울의 음식과 문화 전문가 입니다."),
    ("human", "서울을 대표하는 음식을 맛볼 수 있는 레스토랑 3개를 추천해줘.")
]
```

```python
import asyncio
import time
```

## 동기 vs 비동기 처리

```python
# 동기 처리 하는 함수
def invoke_sync(llm, messages):
    response = llm.invoke(messages)
    print(response.content)
```

```python
# 동기 함수 호출 실행
invoke_sync(llm, messages)
```

```python
# 비동기 처리 하는 함수 정의
async def invoke_async(llm, messages):
    response = await llm.ainvoke(messages) # 비동기 호출, 완료될때까지 기다림(blocking)
    return response.content
```

```python
# 비동기 호출(단일) - 동시성 없음
import time
import asyncio

async def single_async_call():
    start = time.time()
    
    # 비동기 호출이지만, 이 자리에서 완료를 기다림
    result1 = await llm.ainvoke(messages)  # 3초 소요, 여기서 기다림
    print(f"총 시간: {time.time() - start:.2f}초")
    result2 = await llm.ainvoke(messages)  # 3초 소요, 여기서 기다림
    print(f"총 시간: {time.time() - start:.2f}초")
    result3 = await llm.ainvoke(messages)  # 3초 소요, 여기서 기다림
    print(f"총 시간: {time.time() - start:.2f}초")
    print(result1.content)
    print(result2.content)
    print(result3.content)

await single_async_call()
```

## 비동기 처리 방식

```python
# 비동기 처리 하는 함수
async def invoke_async(llm, messages):
    response = await llm.ainvoke(messages)
    print(response.content)

# invoke 병렬처리
async def invoke_parallel(llm, messages):
    tasks = [invoke_async(llm, messages) for _ in range(10)]
    print(tasks)
    
    # 비동기 호출을 동시에 실행
    await asyncio.gather(*tasks)
```

```python
print("비동기 처리")
start = time.perf_counter()
# 병렬처리 코드
await invoke_parallel(llm, messages)
end = time.perf_counter()
print(f"걸린 시간 : {end-start:.2f} secs")
```

# 비동기 처리

```python
print("동기 처리")
# 동기 처리 함수 정의
def invoke_sync(messages):
    # 리스트 컨프리핸션으로 구현
    res = [llm.invoke(messages) for _ in range(2)]
    print(type(res))
    for r in res:
        print(r.content)
start = time.perf_counter()
# 동기 처리  코드
invoke_sync(messages)
end = time.perf_counter()
print(f"걸린 시간 : {end-start:.2f} secs")
```
